### An inspection routine

```python
rng = np.random.default_rng(42)
sales = pd.DataFrame({
    'date': pd.date_range('2025-01-01', periods=100),
    'product': rng.choice(['A', 'B', 'C'], 100),
    'units': rng.integers(50, 500, 100),
    'region': rng.choice(['North', 'South', 'East', 'West'], 100),
})
sales.loc[5:7, 'units'] = np.nan


def inspect_df(df):
    print(f'Shape: {df.shape[0]} rows x {df.shape[1]} columns')
    print('\nFirst 3 rows:')
    print(df.head(3))
    print('\nRandom sample:')
    print(df.sample(3, random_state=42))
    print('\nStructure:')
    df.info()
    print('\nMissing values:')
    print(df.isna().sum())


inspect_df(sales)
```

## 4.7 Setting and Resetting the Index

```python
countries = pd.DataFrame({
    'country': ['USA', 'China', 'India', 'Brazil', 'Germany'],
    'continent': ['North America', 'Asia', 'Asia',
                  'South America', 'Europe'],
    'population_m': [331, 1444, 1393, 213, 83],
    'gdp_trillion': [21.4, 14.7, 2.9, 1.8, 3.8],
})
countries
```

### `df.set_index()`

```python
countries_by_name = countries.set_index('country')
countries_by_name
```

```python
countries_by_name.loc['China']
```

```python
countries.set_index('country', drop=False).head(2)
```

```python
countries_multi = countries.set_index(['continent', 'country'])
countries_multi
```

```python
countries_multi.loc['Asia']
```

### `df.reset_index()`

```python
countries_by_name.reset_index()
```

```python
countries_by_name.reset_index(drop=True).head(2)
```

### Resetting after a filter

```python
asia = countries[countries['continent'] == 'Asia']
asia
```

```python
asia = asia.reset_index(drop=True)
asia
```

### Putting it together

```python
orders = pd.DataFrame({
    'order_id': ['ORD001', 'ORD002', 'ORD003', 'ORD004', 'ORD005'],
    'customer': ['Alice', 'Bob', 'Alice', 'Chen', 'Bob'],
    'amount': [250, 450, 120, 890, 340],
    'status': ['Shipped', 'Pending', 'Shipped', 'Shipped',
               'Cancelled'],
})
orders = orders.set_index('order_id')
print('ORD003 amount:', orders.loc['ORD003', 'amount'])

shipped = orders[orders['status'] == 'Shipped'].reset_index()
shipped
```
