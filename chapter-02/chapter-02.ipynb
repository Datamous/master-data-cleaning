### Returning several values from a function

### Unpacking in loops with `enumerate()` and `zip()`

```python
products = ["Widget A", "Widget B", "Widget C"]
prices = [10.99, 15.50, 8.25]
quantities = [100, 50, 200]

for position, product in enumerate(products):
    print(position, product)

for product, price, qty in zip(products, prices, quantities):
    print(f"{product}: {qty} x {price:.2f} = {price * qty:.2f}")
```

### Skipping and stopping with `continue` and `break`

```python
raw_emails = ["Aisha@Example.com ", "invalid", "", " ben@example.com"]
emails = []

for email in raw_emails:
    email = email.strip()
    if not email or "@" not in email:
        continue
    emails.append(email.lower())

emails
```

```python
required = ["name", "email", "age", "address"]
record = {"name": "Aisha", "email": "aisha@example.com", "age": 30}

missing = None
for field in required:
    if field not in record:
        missing = field
        break

print(f"First missing field: {missing}")
```

## 2.7 Functions and Lambda

### Defining a function with `def`

```python
def add_tax(price, rate):
    return price * (1 + rate)

add_tax(100, 0.06)
```

### Default values

```python
def add_tax(price, rate=0.06):
    return round(price * (1 + rate), 2)

print(add_tax(100))
print(add_tax(100, rate=0.1))
```

### Functions that work on DataFrames

```python
def price_summary(table):
    prices = table["price"]
    return prices.sum(), prices.mean(), prices.max()

total, average, highest = price_summary(inventory)
print(f"Total {total:.2f}, average {average:.2f}, max {highest}")
```

```python
customers = pd.DataFrame({
    "name": ["  Aisha Rahman", "BEN TAN", "chen wei  "],
    "email": ["AISHA@MAIL.COM", " ben@mail.com", "Chen@Mail.com "],
})
customers
```

```python
def clean_text_columns(table, columns):
    cleaned = table.copy()
    for column in columns:
        cleaned[column] = cleaned[column].str.strip().str.lower()
    return cleaned

customers_clean = clean_text_columns(customers, ["name", "email"])
customers_clean
```

### Anonymous functions with `lambda`

```python
sales = pd.DataFrame({
    "employee": ["Aisha", "Ben", "Aisha", "Ben", "Aisha"],
    "amount": [100, 150, 200, 120, 180],
})
sales
```

```python
sales["size"] = sales["amount"].apply(
    lambda x: "Large" if x >= 150 else "Small"
)
sales
```

```python
sales.groupby("employee")["amount"].agg(lambda x: x.max() - x.min())
```

### Building your own cleaning toolkit

```python
def standardize_phone(phone):
    digits = "".join([ch for ch in phone if ch.isdigit()])
    if len(digits) == 11 and digits.startswith("1"):
        digits = digits[1:]
    return digits

raw_phones = ["+1 212-555-0147", "(212) 555-0147", "312.555.0199"]
[standardize_phone(p) for p in raw_phones]
```
