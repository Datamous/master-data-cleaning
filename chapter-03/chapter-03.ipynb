## 3.1 Why NumPy? Performance Over Python Lists

### The loop approach

````python
import numpy as np
import pandas as pd

price_list = [100, 200, 150, 300]
discounted = []
for price in price_list:
    discounted.append(price * 0.9)
discounted
````

````python
price_list * 2
````

````python
prices = np.array([100, 200, 150, 300])
prices * 0.9
````

## 3.6 Boolean Masking and Logic: Filtering Data with Conditions

### Comparison operators

````python
order_values = np.array([45, 120, 80, 200, 150, 30, 90])
order_values > 100
````

````python
order_values == 200
````

### Filtering with a mask

````python
temps = np.array([72, 68, 45, 88, 52, 91, 39, 78])
warm = temps > 70
temps[warm]
````

````python
temps[temps < 50]
````

### Counting matches with `sum()`

````python
print((temps > 70).sum())
print((temps < 50).sum())
````

### Finding positions with `np.where()`

````python
np.where(temps > 80)
````

### Combining conditions

````python
ages = np.array([15, 22, 35, 42, 18, 55, 28, 67, 31])
ages[(ages >= 25) & (ages <= 45)]
````

````python
ages[(ages < 20) | (ages > 60)]
````

````python
working_age = (ages >= 25) & (ages <= 45)
ages[~working_age]
````

````python
sale_amounts = np.array([450, 1200, 850, 200, 1500, 95, 2000, 750])
low_or_high = (sale_amounts < 300) | (sale_amounts > 1000)
sale_amounts[low_or_high & (sale_amounts != 1500)]
````

### `np.any()` and `np.all()`

````python
exam_scores = np.array([65, 72, 88, 95, 78, 82, 91])
print(np.any(exam_scores < 60))
print(np.any(exam_scores > 90))
print(np.all(exam_scores >= 60))
print(np.all(exam_scores == 80))
````

### `np.where()` with three arguments

````python
np.where(temps > 80, 'Hot', 'Moderate')
````

````python
raw_values = np.array([5, 150, 80, 200, 90, 250, 75, 95])
np.where(raw_values > 100, 100, raw_values)
````

````python
shelf_prices = np.array([100, 150, 200, 250, 300])
np.where(shelf_prices > 200, shelf_prices * 0.8, shelf_prices * 0.9)
````

### Replacing placeholder codes with NaN

````python
responses = np.array([5, -999, 3, 4, -999, 2, 5, -999, 4])
cleaned = np.where(responses == -999, np.nan, responses)
cleaned
````

````python
float(np.nanmean(cleaned))
````

### Data cleaning examples

````python
customer_ages = np.array([25, 32, -5, 45, 150, 28, 38, 0, 52, 200, 31])
valid = (customer_ages >= 18) & (customer_ages <= 100)
customer_ages[~valid]
````

````python
invalid_count = (~valid).sum()
share = invalid_count / customer_ages.size * 100
print(f"Invalid ages: {invalid_count} ({share:.1f}%)")
````

````python
item_prices = np.array([15, 25, 50, 75, 100, 150, 200])
item_stock = np.array([0, 50, 100, 5, 80, 0, 120])
available = (item_stock > 0) & (item_prices <= 100)
print(item_prices[available])
print(item_stock[available])
````

### The same idea in pandas

````python
stock_list = pd.DataFrame({
    'product': ['Kettle', 'Toaster', 'Blender', 'Fan'],
    'price': [100, 150, 200, 250],
    'stock': [50, 0, 30, 5],
})
stock_list[(stock_list['stock'] > 0) & (stock_list['price'] <= 200)]
````

## 3.7 Array Aggregation and Summary Statistics

### `sum()`, `mean()`, `min()`, `max()`, `std()` and `var()`

````python
values = np.array([10, 20, 30, 40, 50])
print(f"Sum: {values.sum()}")
print(f"Mean: {values.mean()}")
print(f"Min: {values.min()}")
print(f"Max: {values.max()}")
print(f"Std dev: {values.std():.2f}")
print(f"Variance: {values.var():.2f}")
````

````python
print(np.median(values))
print(values.prod())
````

### `argmin()` and `argmax()`

````python
night_temps = np.array([45, 12, 89, 3, 67, 23])
coldest = night_temps.argmin()
print(f"Coldest at position {coldest}: {night_temps[coldest]}")
hottest = night_temps.argmax()
print(f"Hottest at position {hottest}: {night_temps[hottest]}")
````

### The `axis` parameter

````python
product_sales = np.array([
    [100, 150, 200, 180],
    [120, 130, 140, 150],
    [200, 220, 210, 230],
])
product_sales.sum()
````

````python
product_sales.sum(axis=0)
````

````python
product_sales.sum(axis=1)
````

````python
week_temps = np.array([
    [15, 22, 20, 12],  # Monday
    [16, 24, 22, 14],  # Tuesday
    [14, 21, 19, 11],  # Wednesday
    [17, 25, 23, 15],  # Thursday
    [18, 26, 24, 16],  # Friday
    [19, 27, 25, 17],  # Saturday
    [18, 26, 24, 16],  # Sunday
])
print(f"Daily averages: {week_temps.mean(axis=1)}")
print(f"Time-of-day averages: {week_temps.mean(axis=0).round(1)}")
print(f"Coldest: {week_temps.min()}°C")
print(f"Hottest: {week_temps.max()}°C")
````
